# 04. AI 결과 검수 · 평가

파이프라인이 `data/review/<job_key>/steps_review.csv` 를 내보낸다. 검수자가 채우는 열:

| 열 | 값 |
|---|---|
| `decision` | `approve` / `edit` / `reject` |
| `edited_easy_text` | `edit` 일 때 고친 안내문 |
| `reviewer_note` | 이유 |

채운 파일을 커밋하면 파이프라인이 읽어 승인 단계만 근로자 화면에 올린다 (인계서 5.8).
아래 평가는 기준 답(gold)과 비교해 인계서 10장 지표를 계산한다.

```python
import os, sys
from pathlib import Path
ROOT = Path.cwd().parent            # notebooks/ 에서 실행한다고 가정
sys.path.insert(0, str(ROOT / "backend"))
DATA = ROOT / "data"
import pandas as pd
pd.set_option("display.max_colwidth", 80)

JOB = "kitchen_helper"
rv = pd.read_csv(DATA / f"review/{JOB}/steps_review.csv", dtype=str).fillna("")
rv[["order", "action_source", "easy_text", "easier_text", "needs_review", "checks", "decision"]]
```

```python
# 확인 필요·점검 경고가 있는 단계부터 본다
rv[(rv["needs_review"] == "True") | (rv["checks"] != "")]
```

```python
# 검수 결과 저장 (엑셀로 직접 고쳐도 된다)
rv.to_csv(DATA / f"review/{JOB}/steps_review.csv", index=False, encoding="utf-8-sig")
```
